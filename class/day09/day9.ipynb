## 1. Function Architecture

````python
# Day 9: Functions & Scope

def calculate_summary(numbers: list, scale_factor: float = 1.0):
    if not numbers: return 0, 0.0
    scaled = [n * scale_factor for n in numbers]
    return sum(scaled), sum(scaled) / len(scaled)

total, avg = calculate_summary([10, 20, 30, 40], scale_factor=1.5)
print(f"Scaled Total: {total}, Scaled Average: {avg}")
````
