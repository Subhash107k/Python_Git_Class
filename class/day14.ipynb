## 1. Domain Class Implementation

````python
# Day 14: OOP Basics

class Student:
    school = "Python Academy"
    
    def __init__(self, student_id: str, name: str, gpa: float):
        self.student_id = student_id
        self.name = name
        self.gpa = gpa
        
    def update_gpa(self, new_gpa: float):
        if 0.0 <= new_gpa <= 4.0:
            self.gpa = new_gpa
            
    def __str__(self):
        return f"Student({self.student_id}): {self.name} | GPA: {self.gpa:.2f}"

s = Student("STU101", "Alice", 3.8)
s.update_gpa(3.95)
print(s)
````
