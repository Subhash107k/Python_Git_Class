## 2. Milestone Deliverable — Project 3 Inventory Analytics

````python
# Project 3 Engine
class InventoryAnalytics:
    def __init__(self, prices, quantities):
        self.prices = np.array(prices, dtype=float)
        self.quantities = np.array(quantities, dtype=int)
        
    def total_valuation(self):
        return float(np.sum(self.prices * self.quantities))

engine = InventoryAnalytics([19.99, 49.99, 5.00], [100, 20, 500])
print(f"Project 3 Total Portfolio Value: ${engine.total_valuation():,.2f}")
````
