## 1. Conditional Logic Execution

````python
# Day 4: Grading System Exercise

score = 85.5

if 0 <= score <= 100:
    if score >= 90:
        grade = "A (Distinction)"
    elif score >= 80:
        grade = "B (Very Good)"
    elif score >= 70:
        grade = "C (Good)"
    elif score >= 60:
        grade = "D (Pass)"
    else:
        grade = "F (Fail)"
    print(f"Score: {score} | Grade: {grade}")
else:
    print("Invalid score")
````

## 2. Code Practice & Debugging

````python
# Leap Year Validator
year = 2024
is_leap = (year % 4 == 0 and year % 100 != 0) or (year % 400 == 0)
print(f"Is {year} a leap year? {is_leap}")

# Debugging Drill Fix (Boundary Check)
age = 25
if age >= 18:
    print("Classification: Adult")
elif age >= 13:
    print("Classification: Teenager")
else:
    print("Classification: Child")
````

## 3. More Conditional Examples

````python
# Even or odd check
number = 7
if number % 2 == 0:
    print("Even number")
else:
    print("Odd number")

# Discount eligibility example
age = 16
has_student_id = True
if age >= 18 or has_student_id:
    print("Eligible for discount")
else:
    print("Not eligible")

# Nested condition example
temperature = 32
if temperature > 30:
    print("It's hot")
elif temperature >= 20:
    print("It's warm")
else:
    print("It's cool")
````
