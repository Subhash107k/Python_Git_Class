## 1. Loop Iterations & Control

````python
# Day 5: Loops & Control Flow

print("Counting evens up to 20:")
for num in range(1, 21):
    if num % 2 != 0:
        continue  # Skip odds
    if num > 16:
        break     # Exit early
    print(num, end=" ")
print("\nLoop complete.")
````

## 2. Project 1 Deliverable — Interactive CLI Tool

````python
# Project 1: Personal CLI Assistant v1.0

user_name = "Alex"
print(f"=== PERSONAL CLI ASSISTANT for {user_name} ===")

while True:
    print("\n1. Sum Numbers Utility")
    print("2. String Inverter")
    print("3. Exit")
    choice = "1"  # Simulated input
    
    if choice == "1":
        limit = 10
        print(f"Sum of 1 to {limit}: {sum(range(1, limit + 1))}")
        break
    elif choice == "2":
        print("Reversed: 'Python'[::-1]")
        break
    elif choice == "3":
        print("Goodbye!")
        break
````

## 3. Extra Loop Examples

````python
# Example 1: Print numbers that are multiples of 3
for n in range(1, 16):
    if n % 3 == 0:
        print(n)

# Example 2: Find the first number greater than 20
count = 0
while count <= 20:
    count += 1
    if count > 20:
        print("First number greater than 20:", count)
        break
````
