## 1. Custom Sorting & Tuples

````python
# Day 6: Sequence Operations & Custom Sorting

# Custom Sort Key (Preserved from ListMethod.ipynb)
def proximity_to_fifty(n):
    return abs(n - 50)

numbers = [100, 50, 65, 82, 23]
numbers.sort(key=proximity_to_fifty)
print("Sorted by proximity to 50:", numbers)

# Tuple Unpacking
coordinates = (27.7172, 85.3240, 1400)
lat, lon, elevation = coordinates
print(f"Lat: {lat}, Lon: {lon}, Elev: {elevation}m")
````

## 2. Practice & Git Commands

````python
# Sort words by length practice
words = ["elephant", "dog", "cat", "hippopotamus", "monkey"]
words.sort(key=len)
print("Sorted by word length:", words)
````
