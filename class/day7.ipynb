## 1. Sets & Dictionaries

````python
# Day 7: Sets & Dictionaries

tags_a = {"python", "git", "coding"}
tags_b = {"git", "github", "python"}

print("Union:", tags_a | tags_b)
print("Intersection:", tags_a & tags_b)

# Nested Dictionary
student_db = {
    "STD101": {"name": "Alice", "courses": ["Math", "Python"]},
    "STD102": {"name": "Bob", "courses": ["SQL", "Git"]}
}
print("Alice's Second Course:", student_db["STD101"]["courses"][1])
````
