## 1. Benchmarking & Comprehensions

## 3. Project Example: Nepal Weather Lookup

This small project shows how to store weather information for several districts in Nepal and fetch results from user input.


````python
# Weather project for Nepal districts
weather_data = {
    'Kathmandu': {'temp_c': 22, 'condition': 'Clear', 'humidity': 55},
    'Pokhara': {'temp_c': 24, 'condition': 'Cloudy', 'humidity': 62},
    'Chitwan': {'temp_c': 30, 'condition': 'Sunny', 'humidity': 48},
    'Janakpur': {'temp_c': 28, 'condition': 'Hot', 'humidity': 70},
    'Dhangadhi': {'temp_c': 31, 'condition': 'Sunny', 'humidity': 45},
    'Butwal': {'temp_c': 27, 'condition': 'Rainy', 'humidity': 82},
    'Nepalgunj': {'temp_c': 29, 'condition': 'Warm', 'humidity': 68},
    'Biratnagar': {'temp_c': 26, 'condition': 'Cloudy', 'humidity': 60},
}

def fetch_weather(district):
    district = district.strip().title()
    if district in weather_data:
        info = weather_data[district]
        return f"{district}: {info['temp_c']}°C, {info['condition']}, Humidity {info['humidity']}%"
    return f"Sorry, {district} is not in the sample data."

district_name = input('Enter a district in Nepal: ')
print(fetch_weather(district_name))

# Show all available districts
print('Available districts:', ', '.join(weather_data.keys()))

# Example of a comprehension
sunny_districts = [name for name, info in weather_data.items() if info['condition'] == 'Sunny']
print('Sunny districts:', sunny_districts)
````
