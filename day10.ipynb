## 1. Dynamic Arguments & Lambdas

````python
# Day 10: *args, **kwargs & Lambdas

def audit_logger(event, *scores, **user_info):
    print(f"=== EVENT: {event} ===")
    for k, v in user_info.items(): print(f" - {k}: {v}")
    if scores: print(f"Average Score: {sum(scores)/len(scores):.2f}")

audit_logger("Audit", 88, 92, 79, name="Alice", role="Engineer")

# Map & Filter with Lambda
nums = [1, 2, 3, 4, 5, 6]
evens = list(filter(lambda x: x % 2 == 0, nums))
print("Evens:", evens)
````
