## 1. Inheritance & Encapsulation

````python
# Day 15: Advanced OOP

class BankUser:
    def __init__(self, name: str, user_id: str):
        self.name = name
        self._status = "Active"
        self.__pin = "1234"
        
    def verify_pin(self, pin: str) -> bool:
        return self.__pin == pin

class PremiumUser(BankUser):
    def __init__(self, name: str, user_id: str, points: int):
        super().__init__(name, user_id)
        self.points = points
        
    def display(self):
        print(f"PREMIUM USER: {self.name} | Points: {self.points} | Status: {self._status}")

u = PremiumUser("Alice", "U99", 500)
u.display()
print("PIN Valid:", u.verify_pin("1234"))
````
