## 1. Generators & Decorators

````python
import time

def number_stream(limit):
    curr = 1
    while curr <= limit:
        yield curr
        curr += 1

def timing_decorator(func):
    def wrapper(*args, **kwargs):
        t0 = time.time()
        res = func(*args, **kwargs)
        print(f"[TIMER] {func.__name__} took {time.time()-t0:.6f}s")
        return res
    return wrapper

@timing_decorator
def process_stream(limit):
    return sum(number_stream(limit))

print("Total Sum:", process_stream(100_000))
````
