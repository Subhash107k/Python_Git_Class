## 1. Variables & Type Conversion

````python
# Day 2: Variables & Casting Demonstration

name = "Alice"
age_str = "25"
height_str = "1.75"

# Explicit Type Casting
age = int(age_str)
height = float(height_str)
is_student = True

# Complex Numbers (preserved from existing material)
comp = 3 + 4j
print(f"Complex Value: {comp}, Real: {comp.real}, Imag: {comp.imag}")

print(f"User: {name} (Age: {age}), Height: {height}m, Student: {is_student}")
print(f"Types -> Name: {type(name)}, Age: {type(age)}, Height: {type(height)}")
````

## 2. Interactive Calculator Exercise

````python
# User Calculator Exercise
num1 = float(input("Enter first number: "))
num2 = float(input("Enter second number: "))
result = num1 + num2
print(f"Calculation Result: {num1} + {num2} = {result}")
````

## 3. Code Practice & Debugging

````python
# Temperature Converter Practice
celsius = 25.0
fahrenheit = (celsius * 9/5) + 32
print(f"{celsius}°C is equal to {fahrenheit:.2f}°F")

# Debugging Drill Solution
user_age = int("25")
years_to_hundred = 100 - user_age
print(f"Years until 100: {years_to_hundred}")
````
