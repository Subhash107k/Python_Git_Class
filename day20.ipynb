## 1. Capstone Portfolio Application (`main.py`)

````python
# Day 20 Capstone: Personal Asset & Analytics Management System (PAAMS CLI)
import numpy as np

class Asset:
    def __init__(self, name: str, symbol: str, price: float, quantity: float):
        self.name = name
        self.symbol = symbol
        self.price = price
        self.quantity = quantity
        
    def valuation(self): return self.price * self.quantity
    def __str__(self): return f"{self.name} ({self.symbol}): {self.quantity} units = ${self.valuation():,.2f}"

portfolio = [
    Asset("Bitcoin", "BTC", 65000.0, 1.5),
    Asset("Ethereum", "ETH", 3500.0, 10.0),
    Asset("Apple Inc", "AAPL", 220.0, 50.0)
]

print("=== PAAMS CLI PORTFOLIO SUMMARY ===")
for a in portfolio: print(a)

vals = np.array([a.valuation() for a in portfolio])
print(f"Total Valuation: ${np.sum(vals):,.2f}")
````
