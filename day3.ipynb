## 1. Operators & String Transformations

````python
# Day 3: Operators & String Methods

a, b = 17, 5
print(f"Division: {a / b} | Floor Div: {a // b} | Modulus: {a % b} | Exponent: {a ** b}")

# Preserved String Manipulation Material
raw_phrase = "  python programming is versatile and powerful!  "
clean_phrase = raw_phrase.strip()

print(f"
Cleaned: '{clean_phrase}'")
print(f"Uppercase: {clean_phrase.upper()}")
print(f"Reversed: {clean_phrase[::-1]}")

replaced = clean_phrase.replace("versatile", "awesome")
words = replaced.split()
joined = "-".join(words)
print("Joined with hyphen:", joined)
````

## 2. Code Practice & Debugging

````python
# Palindrome Checker
word = "Racecar".strip().lower()
is_palindrome = word == word[::-1]
print(f"Is '{word}' a palindrome? {is_palindrome}")

# String Immutability Debugging Fix
text = "  python git zero to hero  ".strip().replace(" ", "-")
print("Corrected string transformation:", text)
````
