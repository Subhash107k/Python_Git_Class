## 1. Benchmarking & Comprehensions

````python
import sys

# Preserved benchmark from Day_Two.ipynb
sample_size = 100_000
list_data = list(range(sample_size))
tuple_data = tuple(range(sample_size))

print(f"List Memory : {sys.getsizeof(list_data):,} bytes")
print(f"Tuple Memory: {sys.getsizeof(tuple_data):,} bytes")

# Comprehensions
squares_dict = {x: x**2 for x in range(1, 11) if x % 2 == 0}
print("Even Squares Dict:", squares_dict)
````
