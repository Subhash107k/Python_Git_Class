```python
import matplotlib.pyplot as plt

print("Image shape:", photo.shape)

height, width = photo.shape[:2]

plt.imshow(photo[0:min(200, height), 0:min(300, width)])
plt.axis("off")
plt.show()
```

```python
plt.imshow(photo[::2, ::2])  # Downsampling the image by taking every second pixel  
```

```python
plt.imshow(photo[:,:,0].T)
```

```python
import numpy as np
photo_masked = np.where(photo > 200, 255, 0)  # Masking pixels greater than 200
plt.imshow(photo_masked)    
```
